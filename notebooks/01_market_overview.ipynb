# 01 — Market Overview

Build the European EO market baseline from EUSPA and EARSC figures and document methodology differences.


```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

ROOT = Path("..")
sources = pd.read_csv(ROOT/"data/reference/sources.csv")
print(sources[["publisher","title","publication_date"]].to_string(index=False))

benchmarks = pd.DataFrame([
    ["EUSPA", 772, 13796, 1796, 2026, "EUSPA EO sector scope"],
    ["EARSC", 811, 16036, 2660, 2025, "EARSC industry survey scope"],
], columns=["source","companies","employees","revenue_eur_m","reference_year","scope"])

benchmarks
```

```python
fig, ax = plt.subplots()
ax.bar(benchmarks["source"], benchmarks["revenue_eur_m"])
ax.set_ylabel("Revenue (€m)")
ax.set_title("European EO sector revenue — source-specific benchmarks")
plt.show()
```
