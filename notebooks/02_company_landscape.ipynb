# 02 — Company Landscape

Explore the company universe by technology, value-chain layer, geography and Copernicus relationship.


```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

ROOT = Path("..")
df = pd.read_csv(ROOT/"data/raw/companies_seed_v0_1.csv")
df.head()
```

```python
for col in ["country","primary_layer","technology_domain","maturity_bucket","european_programme_role"]:
    print("\n", col)
    print(df[col].value_counts(dropna=False))
```

```python
ax = df["country"].value_counts().plot(kind="bar")
ax.set_title("Seed company universe by country")
ax.set_ylabel("Companies")
plt.tight_layout()
plt.show()
```
